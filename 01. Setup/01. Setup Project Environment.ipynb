# Set-up the project environment for Formula 1
1. Create External Location databricks-ext-dl1-formula1
2. Create Catalog formula1
3. Create Schemas landing, bronze, silver and gold
4. Create Volume Files in the landing schema

## Access Cloud Storage

```sql
%fs ls 'abfss://formula1@databrickscoursextdl.dfs.core.windows.net/landing'
```

## Create External Location

```sql
-- Note: 'databricks-course-sc' is merely a logical reference in Unity Catalog.
-- In a production environment, this value would be parameterized or injected via CI/CD.
CREATE EXTERNAL LOCATION IF NOT EXISTS databricks_ext_dl_formula1
URL 'abfss://formula1@databrickscoursextdl.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL `databricks-course-sc`)
COMMENT 'External location for the formula1 container';
```

## Create Catalog formula1

```sql
CREATE CATALOG IF NOT EXISTS formula1
MANAGED LOCATION 'abfss://formula1@databrickscoursextdl.dfs.core.windows.net/'
COMMENT 'This is the main catalog for the formula1 project';
```

## Create Schemas landing, bronze, silver and gold

```sql
CREATE SCHEMA IF NOT EXISTS formula1.landing;
CREATE SCHEMA IF NOT EXISTS formula1.bronze
    MANAGED LOCATION 'abfss://formula1@databrickscoursextdl.dfs.core.windows.net/bronze';
CREATE SCHEMA IF NOT EXISTS formula1.silver
    MANAGED LOCATION 'abfss://formula1@databrickscoursextdl.dfs.core.windows.net/bronze';
CREATE SCHEMA IF NOT EXISTS formula1.gold
    MANAGED LOCATION 'abfss://formula1@databrickscoursextdl.dfs.core.windows.net/bronze';
```

```sql
USE CATALOG formula1;
```

```sql
SHOW SCHEMAS;
```

## Create Volume Files

```sql
CREATE EXTERNAL VOLUME formula1.landing.files
LOCATION 'abfss://formula1@databrickscoursextdl.dfs.core.windows.net/landing';
```
